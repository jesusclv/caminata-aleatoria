```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
PASOS = 10000        # Tiros del dado por caminata
CAMINATAS = 1000     # Cuántas veces repetimos el experimento
```

```python
# 1 = Norte, 2 = Oeste, 3 = Sur, 4 = Este
distancias = []
for _ in range(CAMINATAS):
    dados = np.random.randint(1, 5, size=PASOS)
    x = np.sum(dados == 4) - np.sum(dados == 2)
    y = np.sum(dados == 1) - np.sum(dados == 3)
    distancias.append(np.hypot(x, y))

distancias = np.array(distancias)
print(f"Distancia promedio: {distancias.mean():.2f}")
print(f"Desviación estándar: {distancias.std():.2f}")
```

```python
plt.hist(distancias, bins=40, edgecolor='black')
plt.axvline(distancias.mean(), color='red', linestyle='--', label='Promedio')
plt.xlabel('Distancia al origen')
plt.ylabel('Frecuencia')
plt.title(f'{CAMINATAS} caminatas de {PASOS} pasos')
plt.legend()
plt.show()
```
